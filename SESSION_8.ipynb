```python
"""TASK 4:"""
"""

def probability_food_and_dessert(food_users, dessert_users):
    food_set = set(food_users)
    dessert_set = set(dessert_users)

    # Users who ordered both food and dessert
    both_users = food_set.intersection(dessert_set)

    # All unique users in either group
    all_users = food_set.union(dessert_set)

    if len(all_users) == 0:
        return 0

    return len(both_users) / len(all_users)


food_users = [101, 102, 103, 104, 105]
dessert_users = [103, 104, 105, 106, 107]

probability = probability_food_and_dessert(
    food_users, dessert_users
)

print("Probability of ordering both:", probability)

"""
```
