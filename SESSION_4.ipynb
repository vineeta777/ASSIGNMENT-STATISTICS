```python
"""TASK 2:"""
"""
Given data (in thousands): [5, 7, 8, 8, 9, 10, 12, 15, 95]
Step 1: Calculate the mean
Mean = Sum of all values ÷ Number of values
\[
        \text{Mean}=\frac{5+7+8+8+9+10+12+15+95}{9}
\]

\[
        =\frac{169}{9}=\boxed{18.78}
\]

The mean is approximately 18.78 thousand followers.

Step 2: Calculate the median
The data is already arranged in ascending order:

5, 7, 8, 8, 9, 10, 12, 15, 95

There are 9 values, so the median is the 5th value.
\[
        \boxed{\text{Median}=9}
\]

The median is 9 thousand followers.
Step 3: Compare the mean and median
- Mean = 18.78 thousand
- Median = 9 thousand
Answer: The mean is higher than the median, so the data is right-skewed (positively skewed).

Reason: The influencer with 95 thousand followers is an unusually large value that pulls the mean to the right.
"""
```
