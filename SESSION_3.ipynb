```python
"""TASK 2:"""


import statistics

def calculate_standard_deviation(scores):
    return round(statistics.pstdev(scores), 2)

# Example exam scores
scores = [70, 80, 85, 90, 75]

result = calculate_standard_deviation(scores)

print("Exam scores:", scores)
print("Standard deviation:", result)
```

```python
"""TASK 3:"""


import statistics

friend_a = [200, 200, 200, 200, 200]
friend_b = [100, 300, 150, 400, 50]

std_a = statistics.pstdev(friend_a)
std_b = statistics.pstdev(friend_b)

print("Friend A standard deviation:", round(std_a, 2))
print("Friend B standard deviation:", round(std_b, 2))
```

```python
"""TASK 4:"""


import statistics

salaries = [28000, 29000, 27000, 60000, 26500, 27500]

mean_salary = statistics.mean(salaries)
variance = statistics.pvariance(salaries)
std_deviation = statistics.pstdev(salaries)

print("Mean salary:", round(mean_salary, 2))
print("Variance:", round(variance, 2))
print("Standard deviation:", round(std_deviation, 2))
```
