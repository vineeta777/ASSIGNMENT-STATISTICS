```python
"""TASK 4:"""
"""  
# Each row contains [price, quantity_sold]
products = [
    [500, 10],
    [1200, 5],
    [800, 8],
    [1500, 3],
    [300, 15]
]

def calculate_revenue(products):
    revenues = []

    for product in products:
        price = product[0]
        quantity_sold = product[1]

        revenue = price * quantity_sold
        revenues.append(revenue)

    return revenues


# Calculate and print revenue
revenues = calculate_revenue(products)

for i in range(len(products)):
    print("Product", i + 1, "- Revenue: ₹", revenues[i])

"""
```
