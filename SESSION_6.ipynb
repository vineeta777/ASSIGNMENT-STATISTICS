```python
"""TASK 2:"""
"""
State the hypotheses
- \(H_0:\mu_A=\mu_B\) — there is no difference in average order value.
- \(H_1:\mu_A\ne\mu_B\) — there is a difference in average order value.
We use a two-tailed test at a significance level of 0.05.
Step 2: Calculate the t-statistic
\[
    t=\frac{\bar{x}_A-\bar{x}_B}
    {\sqrt{\frac{s_A^2}{n_A}+\frac{s_B^2}{n_B}}}
\]

Substitute the given values:
\[
    t=\frac{350-320}
    {\sqrt{\frac{50^2}{30}+\frac{50^2}{30}}}
\]

\[
    t=\frac{30}{\sqrt{166.67}}
    =\boxed{2.324}
\]

from scipy.stats import ttest_ind_from_stats

# Group A: New home page
mean_a = 350
std_a = 50
n_a = 30

# Group B: Old home page
mean_b = 320
std_b = 50
n_b = 30

# Perform a two-sample, two-tailed t-test
t_stat, p_value = ttest_ind_from_stats(
    mean1=mean_a, std1=std_a, nobs1=n_a,
    mean2=mean_b, std2=std_b, nobs2=n_b,
    equal_var=True
)

print("T-statistic:", round(t_stat, 3))
print("P-value:", round(p_value, 4))

- Significance level: \(\alpha=0.05\)
- P-value: approximately \(0.0236\)
Since the p-value is less than 0.05, reject the null hypothesis.
"""
```
