```python
"""TASK 1:"""
import statistics

# Daily Spotify song play counts
plays = [120, 135, 150, 200, 120, 90, 200]

# Calculate the mean
mean_plays = statistics.mean(plays)

print("Daily song plays:", plays)
print("Mean number of plays:", mean_plays)
```

```python
"""TASK 2:"""


import statistics

# Zomato delivery times in minutes
delivery_times = [30, 25, 40, 35, 30, 45, 30]

# Calculate the median
median_time = statistics.median(delivery_times)

print("Delivery times:", delivery_times)
print("Median delivery time:", median_time, "minutes")

```

```python
"""TASK 3:"""


import statistics

def most_common_rating(ratings):
    return statistics.mode(ratings)

ratings = [5, 4, 4, 3, 5, 4, 2, 4]

result = most_common_rating(ratings)

print("Product ratings:", ratings)
print("Most common rating:", result)
```

```python
"""TASK 4:"""


import statistics

channel_a = [100, 120, 110, 105, 115]
channel_b = [200, 220, 210, 230, 220]
channel_c = [100, 120, 110, 105, 5000]

channels = {
    "Channel A": channel_a,
    "Channel B": channel_b,
    "Channel C": channel_c
}

for name, views in channels.items():
    print("\n", name)
    print("Mean:", statistics.mean(views))
    print("Median:", statistics.median(views))
    print("Mode:", statistics.mode(views))
```

```python
"""TASK 5:"""


import pandas as pd

# Load the Iris dataset from UCI
url = (
    "https://archive.ics.uci.edu/ml/"
    "machine-learning-databases/iris/iris.data"
)

columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]

df = pd.read_csv(url, header=None, names=columns)

# Select one numeric column
data = df["sepal_length"]

# Calculate statistics
print("Mean:", data.mean())
print("Median:", data.median())
print("Mode:", data.mode().tolist())
```
