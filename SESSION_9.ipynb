```python
"""TASK 1:"""
"""

import numpy as np
import matplotlib.pyplot as plt

# Generate 100 random numbers
data = np.random.normal(loc=50, scale=10, size=100)

# Plot histogram
plt.hist(data, bins=10, density=True, alpha=0.6, edgecolor="black")

# Create bell curve
x = np.linspace(20, 80, 200)
mean = 50
std = 10

y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x - mean) / std) ** 2
)

plt.plot(x, y)
plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Normal Distribution (Mean = 50, SD = 10)")
plt.show()

"""
```

```python
"""TASK 2:"""
"""

import numpy as np
from scipy import stats

# Daily step counts for 20 days
steps = [
    4500, 5000, 5500, 6000, 4800,
    5200, 5700, 6100, 4900, 5300,
    5600, 5800, 5100, 5400, 6200,
    5000, 5500, 5900, 4700, 5300
]

mean_value = np.mean(steps)
median_value = np.median(steps)
mode_value = stats.mode(steps, keepdims=False).mode

print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)

"""
```

```python
"""TASK 3:"""
"""

import numpy as np

# Simulate scores for 200 players
np.random.seed(42)
scores = np.random.normal(loc=70, scale=15, size=200)

# Count scores between 55 and 85
within_range = np.sum((scores >= 55) & (scores <= 85))

print("Total players:", len(scores))
print("Players scoring between 55 and 85:", within_range)
print("Percentage:", round(within_range / 200 * 100, 2), "%")

"""
```
