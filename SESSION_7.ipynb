```python
"""TASK 4:"""
"""  

import matplotlib.pyplot as plt
import numpy as np

# Sample dataset
spotify_hours = [1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5, 6]
test_marks = [95, 90, 88, 82, 78, 75, 68, 62, 58, 50]

# Calculate correlation coefficient
correlation = np.corrcoef(spotify_hours, test_marks)[0, 1]
print("Correlation coefficient:", round(correlation, 3))

# Create scatter plot
plt.scatter(spotify_hours, test_marks)

plt.xlabel("Time spent on Spotify (hours/day)")
plt.ylabel("Test marks (out of 100)")
plt.title("Spotify Time vs Test Marks")
plt.grid(True)
plt.show()

"""
```
